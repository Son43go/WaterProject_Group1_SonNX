```python
import os, glob, zipfile, urllib.request
for d in ['data/raw', 'data/processed', 'report', 'sql']: os.makedirs(d, exist_ok=True)
def fetch(url, name):                                   # download once into data/raw
    path = 'data/raw/' + name
    os.makedirs(os.path.dirname(path), exist_ok=True)
    if not os.path.exists(path): print('downloading', url); urllib.request.urlretrieve(url, path)
    return path
def unzip_all(folder='data/raw'):                       # unpack archives, including archives inside archives
    done = set()
    while True:
        todo = [z for z in glob.glob(folder + '/**/*.zip', recursive=True) if z not in done]
        if not todo: break
        for z in todo:
            with zipfile.ZipFile(z) as f: f.extractall(os.path.dirname(z))
            done.add(z)
def find1(name):                                        # locate one file wherever the archive put it
    hits = glob.glob('data/raw/**/' + name, recursive=True); assert hits, 'not found: ' + name; return hits[0]
```

## Bước 1b. Nạp dữ liệu

- **Làm gì:** đọc file gốc, tạo bảng làm việc.
- **Vì sao:** mọi bước sau dùng bảng này, nên số dòng và số cột in ra ở đây là con số ghi vào mục Data của bài.

```python
# Step 1b: Beachwatch water quality samples of the New South Wales Government at 79 swimming sites in Sydney, 1991 to April 2025: 123,530 rows, 10 columns, with daily weather of the city (12,538 rows, 6 columns)
# download (two csv files, 11 MB and 0.5 MB, no account): https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2025/2025-05-20/water_quality.csv  and  .../weather.csv
import pandas as pd, numpy as np, glob
find = lambda n: glob.glob('data/raw/**/' + n, recursive=True)[0]
raw = pd.read_csv(find('water_quality.csv'), parse_dates=['date']); wx = pd.read_csv(find('weather.csv'), parse_dates=['date']); print(raw.shape, wx.shape, raw.date.min().date(), raw.date.max().date())
print(raw.isna().sum()); print(raw.groupby('region').enterococci_cfu_100ml.agg(['size', 'median', lambda s: round(100 * (s > 40).mean(), 1)]))
```

## Bước 2. Làm sạch có nhật ký

- **Làm gì:** kiểm tra miền giá trị, loại dòng lỗi, lưu parquet và nhật ký làm sạch.
- **Vì sao:** phản biện sẽ hỏi vì sao số dòng khác tài liệu gốc; nhật ký là câu trả lời.
- **Kiểm tra:** mở `report/table_cleaning_log.csv`, mỗi dòng bị bỏ phải có lý do.

```python
# Step 2: samples without a bacteria count are removed, one sample per site and day, rain of the same day and of the days before is joined, missing probe readings are flagged; cleaning log
log = [['raw samples', len(raw)]]
df = raw.dropna(subset=['enterococci_cfu_100ml']); log.append(['samples with a bacteria count', len(df)]); df = df.sort_values(['swim_site', 'date']).drop_duplicates(['swim_site', 'date'], keep='last'); log.append(['after keeping one sample per site and day', len(df)])
df['unsafe'] = (df.enterococci_cfu_100ml > 40).astype(int)             # above 40 colony forming units per 100 ml: not rated good for swimming
w = wx.sort_values('date')[['date', 'precipitation_mm', 'max_temp_C']].rename(columns={'precipitation_mm': 'rain_today', 'max_temp_C': 'air_temp_max'})
w['rain_1d'] = w.rain_today.shift(1); w['rain_3d'] = w.rain_today.shift(1).rolling(3).sum(); w['rain_7d'] = w.rain_today.shift(1).rolling(7).sum(); df = df.merge(w, on='date', how='left'); log.append(['samples without weather (removed)', int(df.rain_7d.isna().sum())]); df = df.dropna(subset=['rain_7d'])
df['temp_missing'] = df.water_temperature_c.isna().astype(int); df['cond_missing'] = df.conductivity_ms_cm.isna().astype(int); log += [['samples without water temperature', int(df.temp_missing.sum())], ['samples without conductivity', int(df.cond_missing.sum())]]
df['year'] = df.date.dt.year; df['month'] = df.date.dt.month; df['row_id'] = np.arange(len(df)); df = df.rename(columns={'enterococci_cfu_100ml': 'enterococci', 'water_temperature_c': 'water_temp', 'conductivity_ms_cm': 'conductivity', 'swim_site': 'site'})
log += [['sites', df.site.nunique()], ['unsafe samples (%)', round(100 * df.unsafe.mean(), 1)], ['unsafe samples from 2020 on (%)', round(100 * df[df.year >= 2020].unsafe.mean(), 1)]]
pd.DataFrame(log, columns=['step', 'value']).to_csv('report/table_cleaning_log.csv', index=False); print(log)
df[['row_id', 'region', 'council', 'site', 'date', 'year', 'month', 'latitude', 'longitude', 'enterococci', 'unsafe', 'water_temp', 'conductivity', 'temp_missing', 'cond_missing', 'rain_today', 'rain_1d', 'rain_3d', 'rain_7d', 'air_temp_max']].to_parquet('data/processed/beach.parquet', index=False); print(df.shape)
```

### 2b.1 Nạp bảng sạch và xem năm dòng đầu

- **Làm gì:** đọc file parquet, chọn cột mục tiêu và cột nhóm, lấy mẫu cố định để vẽ nhanh.

> **Nhận xét của nhóm (bắt buộc, một câu có số):** ...

```python
import matplotlib.pyplot as plt
def bar_chart(labels, values, ylabel, path, highlight=None, fmt='{:.1f}'):
    fig, ax = plt.subplots(figsize=(6.5, 3.2))
    colors = ['#F4A261' if i == highlight else '#1B6B6D' for i in range(len(values))]
    bars = ax.bar([str(l) for l in labels], values, color=colors)
    ax.bar_label(bars, labels=[fmt.format(v) for v in values], fontweight='bold', padding=2, fontsize=8 if len(values) > 7 else 10)
    ax.set_ylabel(ylabel); ax.set_ylim(0, max(values) * 1.2)
    ax.spines[['top', 'right']].set_visible(False)          # no top and right spines
    if len(values) > 7: ax.tick_params(axis='x', labelsize=7, rotation=30); plt.setp(ax.get_xticklabels(), ha='right')
    fig.tight_layout(); fig.savefig(path, dpi=300); plt.show()

import pandas as pd, numpy as np
import matplotlib.pyplot as plt
def nice(v, label):                                     # readable numbers on the bars
    v = np.asarray(v, dtype=float); m = np.nanmax(np.abs(v))
    return (v / 1e6, label + ' (million)', '{:.2f}') if m >= 1e6 else (v / 1e3, label + ' (thousand)', '{:.1f}') if m >= 1e4 else (v, label, '{:.2f}' if m < 100 else '{:.0f}')
E = pd.read_parquet('data/processed/beach.parquet'); TGT, GRP = 'unsafe', 'region'
S = E.sample(min(len(E), 50_000), random_state=0)          # a fixed sample keeps every plot fast
NUM = [c for c in S.select_dtypes('number').columns if c != 'row_id' and S[c].nunique() > 1]
print(E.shape[0], 'rows,', E.shape[1], 'columns;', len(NUM), 'numeric columns; target:', TGT, '; group:', GRP); display(E.head(5))
```

### 2b.7 Giá trị ngoại lai

- **Làm gì:** tỷ lệ dòng nằm ngoài 1,5 lần khoảng tứ phân vị của từng cột.
- **Vì sao:** ngoại lai là lỗi đo thì loại và ghi nhật ký; là hiện tượng thật (đỉnh tải, sự cố) thì giữ và nêu trong bài.

> **Nhận xét của nhóm (bắt buộc, một câu có số):** ...

```python
N_ = S[NUM].astype(float); q1, q3 = N_.quantile(.25), N_.quantile(.75); iqr = q3 - q1
out = (100 * ((N_ < q1 - 1.5 * iqr) | (N_ > q3 + 1.5 * iqr)).mean()).sort_values(ascending=False).round(2).rename('outside_1.5_IQR_pct')
out.to_csv('report/table_outliers.csv'); display(out.head(12).to_frame())
```

### 2b.8 Cột nào liên quan tới mục tiêu

- **Làm gì:** tương quan hạng Spearman giữa từng cột số và mục tiêu, vẽ mười cột mạnh nhất.
- **Vì sao:** đây là dự báo sớm về đặc trưng quan trọng; cột tương quan gần 1 có thể là rò rỉ và phải kiểm tra ý nghĩa.

> **Nhận xét của nhóm (bắt buộc, một câu có số):** ...

```python
others = [c for c in NUM if c != TGT]
cor = S[others].astype(float).corrwith(S[TGT].astype(float), method='spearman').dropna(); cor = cor.reindex(cor.abs().sort_values(ascending=False).index).head(10)
cor.rename('spearman_with_target').to_csv('report/table_eda_correlation.csv'); display(cor.round(3).to_frame())
if len(cor) and cor.abs().max() > 0: bar_chart(cor.index, cor.abs().values, 'Absolute rank correlation with ' + TGT, 'report/fig_eda_correlation.png', highlight=0, fmt='{:.2f}')
```

```python
SQL = r"""-- Step 3: sql/queries.sql
CREATE OR REPLACE TABLE bw AS SELECT * FROM 'data/processed/beach.parquet';
-- Q1: samples and share unsafe by five-year period, with the change against the period before (LAG)
SELECT FLOOR(year / 5) * 5 AS period, COUNT(*) AS samples, COUNT(DISTINCT site) AS sites, 100.0 * AVG(unsafe) AS unsafe_pct, 100.0 * AVG(unsafe) - LAG(100.0 * AVG(unsafe)) OVER (ORDER BY FLOOR(year / 5) * 5) AS change_points FROM bw GROUP BY 1 ORDER BY 1;
-- Q2: share unsafe by rain in the three days before the sample
SELECT CASE WHEN rain_3d = 0 THEN '1: no rain' WHEN rain_3d < 5 THEN '2: below 5 mm' WHEN rain_3d < 20 THEN '3: 5 to 20 mm' WHEN rain_3d < 50 THEN '4: 20 to 50 mm' ELSE '5: 50 mm or more' END AS rain_band, COUNT(*) AS samples, 100.0 * AVG(unsafe) AS unsafe_pct, MEDIAN(enterococci) AS median_count FROM bw GROUP BY 1 ORDER BY 1;
-- Q3: sites ranked by the share of unsafe samples since 2015
SELECT site, region, COUNT(*) AS samples, 100.0 * AVG(unsafe) AS unsafe_pct, RANK() OVER (ORDER BY AVG(unsafe) DESC) AS rank FROM bw WHERE year >= 2015 GROUP BY 1, 2 HAVING COUNT(*) >= 100 ORDER BY 5 LIMIT 15;
-- Q4: model table; the history of a site uses only samples taken before the present one
CREATE OR REPLACE TABLE feat AS
SELECT row_id, region, site, date, year, month, latitude, longitude, unsafe, rain_today, rain_1d, rain_3d, rain_7d, air_temp_max, water_temp, conductivity, temp_missing, cond_missing,
       COALESCE(AVG(unsafe) OVER (PARTITION BY site ORDER BY date ROWS BETWEEN UNBOUNDED PRECEDING AND 1 PRECEDING), 0.15) AS site_unsafe_rate_before,
       COALESCE(LN(1 + LAG(enterococci) OVER w), 2.0) AS last_log_count, COALESCE(date_diff('day', LAG(date) OVER w, date), 30) AS days_since_last_sample,
       SIN(2 * PI() * month / 12) AS month_sin, COS(2 * PI() * month / 12) AS month_cos,
       CASE region WHEN 'Sydney Harbour' THEN 0 WHEN 'Northern Sydney' THEN 1 WHEN 'Sydney City' THEN 2 WHEN 'Southern Sydney' THEN 3 ELSE 4 END AS region_code
FROM bw WINDOW w AS (PARTITION BY site ORDER BY date);
SELECT CASE WHEN year < 2020 THEN 'before 2020' ELSE '2020 to 2025' END AS part, COUNT(*) AS n, 100.0 * AVG(unsafe) AS unsafe_pct FROM feat GROUP BY 1 ORDER BY 1 DESC;
"""
open('sql/queries.sql', 'w').write(SQL)
```

## Bước 5. Hình RQ1

- **Làm gì:** vẽ một biểu đồ cột từ kết quả SQL bằng hàm vẽ chuẩn của nhóm.
- **Vì sao:** người đọc phải thấy kết luận của RQ1 trong vài giây.
- **Kiểm tra:** chữ tiếng Anh, không viền trên và phải, số in đậm trên cột, không chữ đè cột. Viết caption nêu kết luận, không mô tả trục.

```python
import matplotlib.pyplot as plt
def bar_chart(labels, values, ylabel, path, highlight=None, fmt='{:.1f}'):
    fig, ax = plt.subplots(figsize=(6.5, 3.2))
    colors = ['#F4A261' if i == highlight else '#1B6B6D' for i in range(len(values))]
    bars = ax.bar([str(l) for l in labels], values, color=colors)
    ax.bar_label(bars, labels=[fmt.format(v) for v in values], fontweight='bold', padding=2, fontsize=8 if len(values) > 7 else 10)
    ax.set_ylabel(ylabel); ax.set_ylim(0, max(values) * 1.2)
    ax.spines[['top', 'right']].set_visible(False)          # no top and right spines
    if len(values) > 7: ax.tick_params(axis='x', labelsize=7, rotation=30); plt.setp(ax.get_xticklabels(), ha='right')
    fig.tight_layout(); fig.savefig(path, dpi=300); plt.show()

fig1 = con.execute("""SELECT CASE WHEN rain_3d = 0 THEN '1: no rain' WHEN rain_3d < 5 THEN '2: below 5 mm' WHEN rain_3d < 20 THEN '3: 5 to 20 mm' WHEN rain_3d < 50 THEN '4: 20 to 50 mm' ELSE '5: 50 mm or more' END AS g, 100.0 * AVG(unsafe) AS v FROM bw GROUP BY 1 ORDER BY 1""").df()
bar_chart(fig1.g, fig1.v, 'Unsafe samples (%)', 'report/fig_rq1.png', highlight=int(fig1.v.idxmax()), fmt='{:.2f}')
```
